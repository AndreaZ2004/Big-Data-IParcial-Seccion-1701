# Propósito de la actividad

En esta actividad se analizará cómo una empresa como **Amazon** utiliza Big Data para optimizar sus servicios, mejorar la experiencia de compra y manejar millones de transacciones diariamente.

El objetivo no es únicamente ejecutar código, sino comprender cómo los datos masivos permiten tomar decisiones en tiempo real dentro de una plataforma de comercio electrónico.


# ¿Qué datos genera?

Amazon genera enormes cantidades de datos diariamente relacionados con las compras y actividades de sus usuarios.

Entre los datos producidos se encuentran:

- Registros de compras
- Búsquedas de productos
- Métodos de pago
- Direcciones de envío
- Reseñas de clientes
- Inventario de productos
- Tiempos de entrega
- Historial de navegación

El volumen de datos es extremadamente grande debido a que millones de personas utilizan Amazon todos los días alrededor del mundo.

La frecuencia de generación de datos es constante y en tiempo real, ya que cada clic, búsqueda o compra produce nueva información.


# ¿Quién los usa?

Los datos de Amazon son utilizados por diferentes actores.

## Empresa

Amazon utiliza los datos para:

- Mejorar sus servicios
- Recomendar productos personalizados
- Optimizar entregas
- Analizar el comportamiento de los clientes

## Analistas y desarrolladores

Los analistas de datos y desarrolladores procesan la información para crear modelos predictivos y mejorar la plataforma.

## Investigadores y gobiernos

En algunos casos, investigadores y entidades gubernamentales pueden utilizar datos estadísticos para estudios, auditorías o regulaciones.

# ¿Qué problema resuelve?

El análisis de datos en Amazon ayuda a resolver múltiples problemas.

Gracias al Big Data, la empresa puede:

- Personalizar recomendaciones de productos
- Reducir tiempos de entrega
- Optimizar inventarios
- Detectar fraudes
- Reducir costos operativos
- Mejorar la experiencia del usuario

Además, permite tomar decisiones rápidas basadas en información en tiempo real.

# ¿Por qué no bastaría Excel?

Excel no sería suficiente para manejar los datos de Amazon debido al enorme volumen y velocidad con la que se generan.

Una hoja de cálculo tradicional presenta limitaciones cuando se manejan millones de registros diarios.

Además, Amazon trabaja con datos estructurados y no estructurados como:

- Imágenes
- Videos
- Reseñas
- Registros de navegación

Por esta razón, se necesitan tecnologías de Big Data capaces de almacenar, procesar y analizar grandes cantidades de información de manera rápida y eficiente.
