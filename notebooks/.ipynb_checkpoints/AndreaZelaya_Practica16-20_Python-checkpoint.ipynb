# Ejercicio 16: DataFrame de Estudiantes

## Enunciado
Crear un DataFrame con información de 10 estudiantes incluyendo:
- Nombre
- Carrera
- Edad
- Promedio
- Ciudad


```python
import pandas as pd

datos_estudiantes = {
    "Nombre": ["Carlos", "María", "José", "Andrea", "Luis", "Fernanda", "Miguel", "Sofía", "Daniel", "Valeria"],
    "Carrera": ["Ingeniería", "Medicina", "Derecho", "Arquitectura", "Contaduría", "Psicología", "Marketing", "Ingeniería", "Diseño", "Enfermería"],
    "Edad": [21, 23, 20, 22, 24, 21, 25, 19, 23, 22],
    "Promedio": [85, 92, 78, 88, 74, 95, 81, 79, 90, 84],
    "Ciudad": ["Tegucigalpa", "San Pedro Sula", "La Ceiba", "Comayagua", "Choluteca", "Danlí", "Juticalpa", "El Progreso", "Puerto Cortés", "Santa Rosa"]
}

df_estudiantes = pd.DataFrame(datos_estudiantes)

print(df_estudiantes.head())

print(df_estudiantes.info())

promedio_general = df_estudiantes["Promedio"].mean()

print(promedio_general)

print(df_estudiantes[df_estudiantes["Promedio"] >= 80])

print(df_estudiantes.sort_values(by="Promedio", ascending=False))
```

## Resultado
- Se creó correctamente el DataFrame de estudiantes.
- Se mostraron las primeras filas del DataFrame.
- Se calculó el promedio general.
- Se filtraron los estudiantes con promedio mayor o igual a 80.
- Los datos fueron ordenados correctamente.

# Ejercicio 17: Análisis de Ventas con pandas

## Enunciado
Crear un DataFrame de ventas con 15 registros incluyendo:
- Fecha
- Vendedor
- Producto
- Categoría
- Cantidad
- Precio unitario

```python
import pandas as pd

ventas = {
    "Fecha": ["2026-01-10", "2026-01-11", "2026-01-12", "2026-01-13", "2026-01-14",
               "2026-01-15", "2026-01-16", "2026-01-17", "2026-01-18", "2026-01-19",
               "2026-01-20", "2026-01-21", "2026-01-22", "2026-01-23", "2026-01-24"],

    "Vendedor": ["Ana", "Luis", "Carlos", "Ana", "María",
                  "Luis", "Carlos", "Ana", "María", "Luis",
                  "Carlos", "Ana", "María", "Luis", "Carlos"],

    "Producto": ["Laptop", "Mouse", "Teclado", "Monitor", "Laptop",
                  "Impresora", "Mouse", "Tablet", "Monitor", "Laptop",
                  "Teclado", "Tablet", "Mouse", "Impresora", "Monitor"],

    "Categoria": ["Tecnología", "Accesorios", "Accesorios", "Tecnología", "Tecnología",
                   "Oficina", "Accesorios", "Tecnología", "Tecnología", "Tecnología",
                   "Accesorios", "Tecnología", "Accesorios", "Oficina", "Tecnología"],

    "Cantidad": [2, 5, 3, 2, 1, 2, 6, 2, 1, 3, 4, 1, 5, 2, 2],

    "PrecioUnitario": [18000, 350, 900, 4500, 20000,
                        6000, 350, 12000, 5000, 19000,
                        950, 12500, 400, 6200, 4800]
}

df_ventas = pd.DataFrame(ventas)

df_ventas["TotalVenta"] = df_ventas["Cantidad"] * df_ventas["PrecioUnitario"]

print(df_ventas)

print(df_ventas.groupby("Vendedor")["TotalVenta"].sum())

producto_mas_vendido = df_ventas.groupby("Producto")["Cantidad"].sum()

print(producto_mas_vendido.sort_values(ascending=False))

categoria_ingreso = df_ventas.groupby("Categoria")["TotalVenta"].sum()

print(categoria_ingreso.sort_values(ascending=False))

print(df_ventas.sort_values(by="TotalVenta", ascending=False))
```

## Resultado
- Se creó el DataFrame de ventas correctamente.
- Se calculó la columna TotalVenta.
- Se obtuvo el total vendido por cada vendedor.
- Se identificó el producto más vendido.
- Se encontró la categoría con mayor ingreso.

# Ejercicio 18: Limpieza de Datos con pandas

## Enunciado
Crear un DataFrame con datos incompletos de clientes incluyendo:
- Nombre
- Edad
- Correo
- Ciudad
- Teléfono

```python
import pandas as pd

clientes = {
    "Nombre": ["Laura", "Pedro", "Camila", "Andrés", "Valentina"],
    "Edad": [25, 30, None, 28, 35],
    "Correo": ["laura@gmail.com", None, "camila@gmail.com", "andres@gmail.com", None],
    "Ciudad": ["Tegucigalpa", "", "San Pedro Sula", "", "La Ceiba"],
    "Telefono": ["9999-1111", "8888-2222", None, "7777-3333", "6666-4444"]
}

df_clientes = pd.DataFrame(clientes)

print(df_clientes.isnull())

print(df_clientes.isnull().sum())

df_clientes["Ciudad"] = df_clientes["Ciudad"].replace("", "No especificada")

df_clientes_limpio = df_clientes.dropna(subset=["Correo"])

print(df_clientes_limpio)
```

## Resultado
- Se identificaron los valores nulos del DataFrame.
- Se contaron los valores faltantes por columna.
- Las ciudades vacías fueron reemplazadas correctamente.
- Se eliminaron registros sin correo electrónico.

# Ejercicio 19: Análisis de Préstamos

## Enunciado
Crear un DataFrame con solicitudes de préstamo incluyendo:
- Cliente
- Monto solicitado
- Plazo
- Ingreso mensual
- Estado de solicitud

```python
import pandas as pd

prestamos = {
    "Cliente": ["Mario", "Lucía", "José", "Andrea", "Fernando", "Gabriela", "Kevin", "Paola", "Ricardo", "Elena"],
    "MontoSolicitado": [45000, 70000, 30000, 85000, 60000, 25000, 95000, 40000, 75000, 55000],
    "Plazo": [12, 24, 6, 36, 24, 12, 48, 18, 36, 24],
    "IngresoMensual": [18000, 35000, 15000, 42000, 28000, 12000, 50000, 20000, 38000, 26000],
    "Estado": ["Aprobado", "Pendiente", "Rechazado", "Aprobado", "Pendiente", "Rechazado", "Aprobado", "Pendiente", "Aprobado", "Aprobado"]
}

df_prestamos = pd.DataFrame(prestamos)

print(df_prestamos["MontoSolicitado"].mean())

print(df_prestamos["Estado"].value_counts())

print(df_prestamos[df_prestamos["Estado"] == "Aprobado"])

print(df_prestamos[df_prestamos["MontoSolicitado"] > 50000])

print(df_prestamos.groupby("Estado")["MontoSolicitado"].mean())
```

## Resultado
- Se calculó el promedio de préstamos solicitados.
- Se contaron las solicitudes por estado.
- Se mostraron únicamente los préstamos aprobados.
- Se identificaron clientes con montos mayores a 50,000.

# Ejercicio 20: Mini Análisis de Datos Tipo Empresa

## Enunciado
Crear un DataFrame con información de empleados incluyendo:
- Nombre
- Departamento
- Cargo
- Salario
- Años de experiencia
- Ciudad

```python
import pandas as pd

empleados = {
    "Nombre": ["Carlos", "María", "José", "Andrea", "Luis", "Fernanda", "Miguel", "Sofía", "Daniel", "Valeria"],
    "Departamento": ["Tecnología", "Ventas", "Recursos Humanos", "Marketing", "Finanzas", "Tecnología", "Ventas", "Marketing", "Finanzas", "Tecnología"],
    "Cargo": ["Programador", "Supervisor", "Asistente", "Diseñadora", "Contador", "Analista", "Vendedor", "Community Manager", "Auditor", "Desarrolladora"],
    "Salario": [28000, 35000, 18000, 26000, 32000, 30000, 22000, 24000, 34000, 40000],
    "Experiencia": [2, 7, 1, 4, 6, 5, 3, 2, 8, 10],
    "Ciudad": ["Tegucigalpa", "San Pedro Sula", "La Ceiba", "Comayagua", "Danlí", "Juticalpa", "Choluteca", "El Progreso", "Puerto Cortés", "Santa Rosa"]
}

df_empleados = pd.DataFrame(empleados)

print(df_empleados.head())

print(df_empleados["Salario"].describe())

print(df_empleados.groupby("Departamento")["Salario"].mean())

print(df_empleados[df_empleados["Salario"] == df_empleados["Salario"].max()])

print(df_empleados[df_empleados["Experiencia"] > 5])

print(df_empleados.sort_values(by="Salario", ascending=False))

def nivel_experiencia(anios):
    if anios <= 2:
        return "Junior"
    elif anios <= 5:
        return "Intermedio"
    else:
        return "Senior"

df_empleados["NivelExperiencia"] = df_empleados["Experiencia"].apply(nivel_experiencia)

print(df_empleados)
```

## Resultado
- Se creó correctamente el DataFrame de empleados.
- Se calcularon estadísticas generales de salarios.
- Se obtuvo el salario promedio por departamento.
- Se identificó el empleado con mayor salario.
- Se creó correctamente la columna NivelExperiencia.
