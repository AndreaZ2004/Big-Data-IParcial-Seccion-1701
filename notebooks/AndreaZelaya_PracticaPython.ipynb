# Ejercicio 1: Clasificación de edad
## Enunciado:
Determinar si una persona es niño, adolescente, adulto o adulto mayor según su edad.


```python
def clasificar_edad(edad):
    if edad < 13:
        return "Niño"
    elif edad <= 17:
        return "Adolescente"
    elif edad <= 59:
        return "Adulto"
    else:
        return "Adulto mayor"

edad = 21
categoria = clasificar_edad(edad)

print(f"Edad: {edad} años")
print(f"Categoría: {categoria}")
```

## Explicación breve:
Se utiliza una función con condiciones if, elif y else para clasificar la edad según el rango correspondiente.

# Ejercicio 2: Conversor de temperatura
## Enunciado:
Convertir una temperatura de grados Celsius a Fahrenheit.

```python

def convertir_temperatura(celsius):
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit

celsius = 30
fahrenheit = convertir_temperatura(celsius)

print(f"Temperatura en Celsius: {celsius}°C")
print(f"Temperatura en Fahrenheit: {fahrenheit}°F")
```

## Explicación breve:
Se aplica la fórmula de conversión de Celsius a Fahrenheit utilizando una función para retornar el resultado.

# Ejercicio 3: Calculadora básica

## Enunciado:
Realizar operaciones matemáticas básicas y validar división entre cero.

```python

def calculadora(num1, num2, operacion):
    if operacion == "suma":
        return num1 + num2
    elif operacion == "resta":
        return num1 - num2
    elif operacion == "multiplicacion":
        return num1 * num2
    elif operacion == "division":
        if num2 != 0:
            return num1 / num2
        else:
            return "Error: No se puede dividir entre cero"
    else:
        return "Operación no válida"

num1 = 10
num2 = 5
operacion = "division"

resultado = calculadora(num1, num2, operacion)

print(f"Resultado: {resultado}")
```

## Explicación breve:
El programa utiliza condiciones para identificar la operación matemática seleccionada y validar que no exista división entre cero.

# Ejercicio 5: Tabla de multiplicar
## Enunciado:
Mostrar la tabla de multiplicar del 1 al 12.

```python
numero = 7

print(f"Tabla de multiplicar del {numero}")

for i in range(1, 13):
    resultado = numero * i
    print(f"{numero} x {i} = {resultado}")
```

# Ejercicio 6: Contador de vocales
## Enunciado
Contar cuántas vocales contiene una frase.

```python
frase = "Python es un lenguaje de programacion"

vocales = "aeiouAEIOU"
contador = 0

for letra in frase:
    if letra in vocales:
        contador += 1

print(f"Frase: {frase}")
print(f"Cantidad de vocales: {contador}")
```

# Ejercicio 7: Promedio de calificaciones
## Enunciado:
Calcular el promedio de 5 calificaciones e indicar si aprobó.

```python
calificaciones = [85, 90, 78, 88, 95]

suma = sum(calificaciones)
promedio = suma / len(calificaciones)

print(f"Calificaciones: {calificaciones}")
print(f"Suma total: {suma}")
print(f"Promedio: {promedio}")

if promedio >= 70:
    print("Estado: Aprobado")
else:
    print("Estado: Reprobado")
```

## Explicación breve:
 Se utiliza una lista para almacenar las calificaciones,
luego se calcula la suma y el promedio.

# Ejercicio 8: Lista de números positivos y negativos
## Enunciado:
 Contar números positivos, negativos y ceros.

```python
numeros = [10, -5, 0, 8, -2, 15, 0, -7, 20, 3]

positivos = 0
negativos = 0
ceros = 0

for numero in numeros:
    if numero > 0:
        positivos += 1
    elif numero < 0:
        negativos += 1
    else:
        ceros += 1

print(f"Positivos: {positivos}")
print(f"Negativos: {negativos}")
print(f"Ceros: {ceros}")
```

## Explicación breve:
Se recorre la lista y se clasifica cada número dependiendo de su valor.

# Ejercicio 11: Función para calcular descuento
## Enunciado:
Calcular el precio final aplicando descuento.

```python
def calcular_descuento(precio_original, porcentaje_descuento):
    descuento = precio_original * (porcentaje_descuento / 100)
    precio_final = precio_original - descuento
    return precio_final

precio = 1500
descuento = 20

resultado = calcular_descuento(precio, descuento)

print(f"Precio final: L.{resultado}")
```

## Explicación breve:
La función recibe el precio y el porcentaje de descuento, luego retorna el precio final.

# Ejercicio 12: Validación de contraseña
## Enunciado:
Validar si una contraseña cumple los requisitos.

```python
contraseña = "Python2026"

tiene_numero = False
tiene_mayuscula = False
tiene_minuscula = False

for caracter in contraseña:
    if caracter.isdigit():
        tiene_numero = True
    if caracter.isupper():
        tiene_mayuscula = True
    if caracter.islower():
        tiene_minuscula = True

if len(contraseña) >= 8 and tiene_numero and tiene_mayuscula and tiene_minuscula:
    print("La contraseña es válida")
else:
    print("La contraseña no es válida")
```

## Explicación breve:
Se verifican los requisitos de longitud, números, mayúsculas y minúsculas.

# Ejercicio 13: Sistema simple de inventario
## Enunciado:
Crear un sistema básico de inventario.

```python
productos = [
    {"codigo": 1, "nombre": "Laptop", "precio": 25000, "cantidad": 5},
    {"codigo": 2, "nombre": "Mouse", "precio": 500, "cantidad": 20},
    {"codigo": 3, "nombre": "Teclado", "precio": 1200, "cantidad": 10}
]

for producto in productos:
    print(producto)

valor_total = 0

for producto in productos:
    valor_total += producto["precio"] * producto["cantidad"]

print(f"Valor total del inventario: L.{valor_total}")
```

## Explicación breve:
Se utilizan listas y diccionarios para almacenar la información de cada producto.

# Ejercicio 14: Registro de ventas
## Enunciado:
Registrar ventas y calcular totales.

```python
ventas = [
    {"cliente": "Juan", "producto": "Laptop", "cantidad": 1, "precio_unitario": 25000},
    {"cliente": "Maria", "producto": "Mouse", "cantidad": 3, "precio_unitario": 500},
    {"cliente": "Carlos", "producto": "Teclado", "cantidad": 2, "precio_unitario": 1200}
]

total_general = 0

for venta in ventas:
    total_venta = venta["cantidad"] * venta["precio_unitario"]
    total_general += total_venta

    print(f"Cliente: {venta['cliente']}")
    print(f"Total venta: L.{total_venta}")

print(f"Total general vendido: L.{total_general}")
```

## Explicación breve:
Se calcula el total de cada venta multiplicando cantidad por precio unitario.

# Ejercicio 15: Limpieza básica de texto
## Enunciado:
Limpiar espacios y analizar una frase.

```python
frase = "   Python   es   un lenguaje   poderoso   "

texto_limpio = frase.strip()
texto_limpio = texto_limpio.lower()
texto_limpio = texto_limpio.replace("   ", " ")
texto_limpio = texto_limpio.replace("  ", " ")

palabras = texto_limpio.split()
cantidad_palabras = len(palabras)

print(f"Texto limpio: '{texto_limpio}'")
print(f"Cantidad de palabras: {cantidad_palabras}")
```

## Explicación breve:
Se utilizan métodos de texto para limpiar espacios, convertir a minúsculas y dividir palabras.
